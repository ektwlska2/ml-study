## Part B · On paper

**B1.** A node holds 6 samples: 4 of class 1 and 2 of class 0. A split sends 3 samples (all class 1) left and 3 samples (1 of class 1, 2 of class 0) right. Compute the parent's Gini, each child's Gini, and the Gini decrease.

<details><summary>Answer</summary>

Parent: $p = \frac{2}{3}$, Gini $= 2 \cdot \frac{2}{3} \cdot \frac{1}{3} = 0.444$. Left: pure, Gini $0$. Right: $p = \frac{1}{3}$, Gini $= 0.444$. Weighted: $\frac{3}{6} \times 0 + \frac{3}{6} \times 0.444 = 0.222$. Decrease: $0.444 - 0.222 = 0.222$.

</details>

**B2.** Linear SVM. One negative at $(1, 0)$ and one positive at $(3, 0)$. Find $w$, $b$ (with $yz = 1$ at both points) and the margin width. Then give the hinge loss of a new positive point at $(2.5, 0)$.

<details><summary>Answer</summary>

The line $x_1 = 2$ is midway between them, so $w = (c, 0)$. $c + b = -1$ and $3c + b = 1$, so $c = 1$ and $b = -2$: $w = (1, 0)$, $b = -2$, width $\frac{2}{1} = 2$. For $(2.5, 0)$: $z = 0.5$, $yz = 0.5$, hinge $= 1 - 0.5 = 0.5$ (inside the street, right side).

</details>

**B3.** AdaBoost, round 1: 10 points with equal weights, and the chosen stump misclassifies 3 of them. Compute $\varepsilon$, $\alpha$, and the new weight of a misclassified point and of a correct one (after normalizing).

<details><summary>Answer</summary>

$\varepsilon = 0.3$. $\alpha = \frac{1}{2}\ln\frac{0.7}{0.3} = 0.424$. Before normalizing: wrong $0.1 \times e^{0.424} = 0.1528$, right $0.1 \times e^{-0.424} = 0.0655$. Sum: $3 \times 0.1528 + 7 \times 0.0655 = 0.917$. After: **wrong $0.167$, right $0.0714$**. The three mistakes hold $3 \times 0.167 = 0.5$ of the total.

</details>

**B4.** A detector ranks 6 parts; the true labels in rank order are $1, 1, 0, 1, 0, 0$. Compute the precision at each rank and the average precision. What would random ranking give, roughly?

<details><summary>Answer</summary>

Precision: $1, 1, 0.667, 0.75, 0.6, 0.5$. Recall rises at ranks 1, 2 and 4, so $\text{AP} = \frac{1}{3}(1 + 1 + 0.75) = 0.917$. Random ranking gives about the share of positives, $\frac{3}{6} = 0.5$.

</details>
