**Drill (on paper).** **(1. Gini and information gain.)** On the table above, compute the Gini decrease and the information gain of splitting on $A$ and on $B$. Which feature goes at the root?

<details><summary>Worked answer</summary>

Parent: 3 of 8 positive. Gini $= 2 \cdot \frac{3}{8} \cdot \frac{5}{8} = 0.469$; entropy $= 0.954$. **Split on $A$:** $A = 1$ has labels $1, 1, 1, 0$, so Gini $= 2 \cdot 0.75 \cdot 0.25 = 0.375$ and $H = 0.811$. $A = 0$ is all 0: Gini $= H = 0$. Weighted Gini $= 0.5 \times 0.375 = 0.188$, so the decrease is $0.469 - 0.188 = 0.281$. Weighted $H = 0.406$, so the gain is $0.954 - 0.406 = 0.549$. **Split on $B$:** $B = 1$ (samples 1, 3, 5, 7) has labels $1, 1, 0, 0$: Gini $0.5$, $H = 1$. $B = 0$ (samples 2, 4, 6, 8) has $1, 0, 0, 0$: Gini $0.375$, $H = 0.811$. Weighted Gini $0.438$, decrease $0.031$; weighted $H = 0.906$, gain $0.049$. **Root: $A$**, by both criteria.

</details>

**Drill (on paper).** **(Pruning.)** A subtree has 5 leaves and training error 0.04. Replaced by a single leaf, the error is 0.12. For which $\alpha$ does cost-complexity pruning remove the subtree?

<details><summary>Worked answer</summary>

Keep it while $0.04 + 5\alpha < 0.12 + \alpha$, i.e. $4\alpha < 0.08$, i.e. $\alpha < 0.02$. **Pruned for $\alpha > 0.02$** (at exactly 0.02 the two are tied).

</details>

**Drill (on paper).** **(Handwritten code.)** Write `gini(y)` and `best_threshold(x, y)` in NumPy: try every midpoint between sorted unique values, and return the threshold with the largest Gini decrease.

<details><summary>Worked answer</summary>

```python
def gini(y):
    p = np.bincount(y) / len(y)
    return 1 - np.sum(p ** 2)

def best_threshold(x, y):
    v = np.unique(x)
    best_t, best_g = None, 0.0
    for t in (v[:-1] + v[1:]) / 2:
        left = x <= t
        g = gini(y) - left.mean() * gini(y[left]) - (~left).mean() * gini(y[~left])
        if best_t is None or g > best_g:
            best_t, best_g = t, g
    return best_t, best_g
```
`left.mean()` is $\frac{n_L}{n}$: the mean of a True/False array is the fraction of Trues. Compare with your exercises 1 and 4 in [04 §8 ↗](04_decision_trees.ipynb).

</details>

## B. SVM

**Drill (on paper).** **(Hard margin, by hand.)** Negatives $(0, 0)$ and $(0, 1)$; positives $(2, 0)$, $(2, 1)$ and $(3, 1)$. Find the maximum-margin line ($w$, $b$, scaled so the support vectors have $y z = 1$), the margin width, and the support vectors.

<details><summary>Worked answer</summary>

The negatives sit at $x_1 = 0$ and the closest positives at $x_1 = 2$, so the widest street is vertical, centred at $x_1 = 1$: $w = (c, 0)$. Scores: $c \cdot 0 + b = -1$ and $c \cdot 2 + b = +1$, so $b = -1$ and $c = 1$. **$w = (1, 0)$, $b = -1$**, width $\frac{2}{\lVert w \rVert} = 2$. Support vectors: $(0, 0)$, $(0, 1)$, $(2, 0)$, $(2, 1)$, all with $yz = 1$. $(3, 1)$ has $yz = 2$, so it is **not** a support vector.

</details>

**Drill (on paper).** **(Two points.)** One negative at $(0, 0)$ and one positive at $(2, 2)$. Find $w$, $b$ and the margin.

<details><summary>Worked answer</summary>

The line is the perpendicular bisector of the two points: through $(1, 1)$, perpendicular to $(2, 2)$, so $w = c\,(1, 1)$. $w^\top (0, 0) + b = -1$ gives $b = -1$; $w^\top (2, 2) + b = 4c - 1 = 1$ gives $c = 0.5$. **$w = (0.5, 0.5)$, $b = -1$**, $\lVert w \rVert = 0.707$, width $= 2.828$, which is exactly the distance between the two points, as it should be.

</details>

**Drill (on paper).** **(Hinge loss.)** $w = (1, -1)$, $b = 0$. For the points $\big((2, 0), +1\big)$, $\big((1, 0.5), +1\big)$, $\big((0, 1), +1\big)$ and $\big((0, 2), -1\big)$, compute $yz$, the hinge loss, and say which points pull on $w$.

<details><summary>Worked answer</summary>

$z = x_1 - x_2$. $(2, 0)$: $z = 2$, $yz = 2$, hinge $0$, no pull. $(1, 0.5)$: $z = 0.5$, $yz = 0.5$, hinge $0.5$: inside the street, **pulls**. $(0, 1)$: $z = -1$, $yz = -1$, hinge $2$: wrong side, **pulls**. $(0, 2)$, label $-1$: $z = -2$, $yz = 2$, hinge $0$, no pull. Pulling points push $w$ toward $y_i x_i$: here toward $(1, 0.5)$ and toward $(0, 1)$.

</details>

**Drill (on paper).** **(Kernel.)** $x = (1, 1)$, $z = (2, -1)$. Compute $k(x, z) = (x^\top z + 1)^2$ directly, then again as $\phi(x)^\top \phi(z)$ with $\phi(v) = (1, \sqrt2 v_1, \sqrt2 v_2, v_1^2, \sqrt2 v_1 v_2, v_2^2)$.

<details><summary>Worked answer</summary>

Directly: $x^\top z = 2 - 1 = 1$, so $(1 + 1)^2 = 4$. $\phi(x) = (1, \sqrt2, \sqrt2, 1, \sqrt2, 1)$ and $\phi(z) = (1, 2\sqrt2, -\sqrt2, 4, -2\sqrt2, 1)$. Products: $1 + 4 - 2 + 4 - 4 + 1 = 4$ ✓.

</details>

**Drill (on paper).** **(Dual, conceptual.)** In the dual SVM, $w = \sum_i \alpha_i y_i x_i$. Which $\alpha_i$ are zero, and why does the dual only need $x_i^\top x_j$?

<details><summary>Worked answer</summary>

$\alpha_i = 0$ for every point strictly outside the margin ($y_i z_i > 1$); only support vectors have $\alpha_i > 0$ (complementary slackness). Substituting $w = \sum_i \alpha_i y_i x_i$ into the Lagrangian leaves $\sum_i \alpha_i - \frac{1}{2}\sum_{i,j} \alpha_i \alpha_j y_i y_j\, x_i^\top x_j$, and a prediction is $\sum_i \alpha_i y_i\, x_i^\top x + b$. Points only ever appear inside dot products, so a kernel can replace them. [05 §7 ↗](05_svm.ipynb)

</details>

**Drill (on paper).** **(AdaBoost, one round.)** $x = 1, 2, 3, 4, 5$ with labels $+1, +1, -1, +1, -1$, equal weights $0.2$. Stumps: "predict $s$ if $x > t$, else $-s$", with the leftmost threshold winning ties. Find the first stump, $\varepsilon$, $\alpha$ and the new weights.

<details><summary>Worked answer</summary>

Best stumps: $t = 2.5$, $s = -1$ (predict $+$ for $x \le 2.5$, $-$ above) misclassifies only $x = 4$, so $\varepsilon = 0.2$. ($t = 4.5$, $s = -1$ also gives 0.2, but 2.5 comes first.) $\alpha = \frac{1}{2}\ln\frac{0.8}{0.2} = \frac{1}{2}\ln 4 = 0.693$. Correct points: $0.2 \times e^{-0.693} = 0.1$. Wrong point: $0.2 \times e^{0.693} = 0.4$. Sum $0.8$. **New weights: $0.125, 0.125, 0.125, 0.5, 0.125$.** The single mistake now holds half of the total weight.

</details>

**Drill (on paper).** **(Variance of an average.)** 20 trees, each with prediction variance 4 and pairwise correlation 0.5. Variance of their average? What is the lowest it could go with infinitely many trees?

<details><summary>Worked answer</summary>

$\rho\sigma^2 + \frac{1 - \rho}{B}\sigma^2 = 0.5 \times 4 + \frac{0.5 \times 4}{20} = 2 + 0.1 = 2.1$. With $B \to \infty$: $\rho\sigma^2 = 2$. Only decorrelating the trees (random forests) gets below 2.

</details>

**Drill (on paper).** **(Average precision.)** Five items ranked by score; the real labels in that order are $1, 0, 1, 0, 1$. Compute precision at each rank, the AP, and the ROC-AUC.

<details><summary>Worked answer</summary>

Precision at ranks 1–5: $1,\ 0.5,\ 0.667,\ 0.5,\ 0.6$. Recall rises at ranks 1, 3 and 5. $\text{AP} = \frac{1}{3}(1 + 0.667 + 0.6) = 0.756$. ROC-AUC: the positive at rank 1 beats both negatives, the one at rank 3 beats one, the one at rank 5 beats none: $\frac{3}{6} = 0.5$. Same ranking, very different numbers: AP rewards getting the **top** of the list right, while ROC-AUC counts all pairs equally.

**Drill (on paper).** **(Handwritten code.)** Write `knn_predict(X_train, y_train, X_test, k)` for integer labels, with no loop over training points.

<details><summary>Worked answer</summary>

```python
def knn_predict(X_train, y_train, X_test, k):
    D = np.sqrt(((X_test[:, None, :] - X_train[None, :, :]) ** 2).sum(axis=2))   # (m, n)
    idx = np.argsort(D, axis=1)[:, :k]                                           # (m, k)
    return np.array([np.bincount(row).argmax() for row in y_train[idx]])         # (m,)
```
