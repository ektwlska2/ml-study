## Words you'll see
Come back to this list whenever a word is unclear.

- **value**: A piece of data, like `7`, `2.5`, or `"hello"`.
- **variable**: A name you give a value so you can use it later. `x = 7` means "from now on, `x` means 7".
- **type**: What kind of value something is: a whole number, a decimal, text, and so on.
- **function**: A named piece of code that does one job. You *call* it with parentheses: `len("hi")`.
- **argument**: The input you put inside the parentheses when you call a function. In `len("hi")`, the argument is `"hi"`.
- **return value**: The output a function gives back. `len("hi")` returns `2`.
- **method**: A function that belongs to a value. You call it with a dot: `"hi".upper()` returns `"HI"`.
- **index**: The position number of an item. **Python counts from 0**, so the first item is at index 0.
- **loop**: Code that repeats, for example once for every item in a list.
- **error**: Python stopping because something is wrong. The last line of the message tells you what (e.g. `TypeError`, `IndexError`). Reading it is a skill, not a failure.

## 1. Variables and numbers

`=` in Python does **not** mean "equals" like in math. It means: *work out the right side, then store the result under the name on the left.*

```python
price = 3000            # store 3000 under the name price
count = 4               # store 4 under the name count
total = price * count   # Python computes 3000 * 4 = 12000, stores it in total
print(total)            # → 12000
```

You can change a variable later. The old value is simply replaced:

```python
count = 4
count = count + 1       # right side first: 4 + 1 = 5, then store 5 in count
print(count)            # → 5
```

Every value has a **type**. You only need these five for now:

| Type | What it is | Examples |
|---|---|---|
| `int` | whole number | `7`, `-3`, `0` |
| `float` | number with a decimal point | `2.5`, `0.1`, `3.0` |
| `str` | text ("string"), always inside quotes | `"hello"`, `'Kim'`, `"42"` |
| `bool` | yes/no | `True`, `False` |
| `None` | "no value here" | `None` |

```python
print(type(7))        # → <class 'int'>
print(type(2.5))      # → <class 'float'>
print(type("hello"))  # → <class 'str'>
```

The four arithmetic signs that confuse people most, shown with 7 and 2:

| Write | Result | Meaning |
|---|---|---|
| `7 / 2` | `3.5` | normal division (the result is always a float) |
| `7 // 2` | `3` | how many *whole* times 2 fits into 7 |
| `7 % 2` | `1` | what's *left over* after that |
| `2 ** 3` | `8` | power: 2 × 2 × 2 |

```python
minutes = 200
hours = minutes // 60       # 60 fits into 200 three whole times → 3
leftover = minutes % 60     # 200 - 3*60 = 20 left over → 20
print(hours, leftover)      # → 3 20
```

**Text that looks like a number is still text.** `"3"` is a string, not the number 3. Use `int()` or `float()` to convert:

```python
print(3 + 4)              # → 7      (numbers: added)
print("3" + "4")          # → 34     (strings: glued together)
print(int("3") + 4)       # → 7      (converted to a number first)
```

An **f-string** puts values inside text. Write `f` before the quotes and put names in `{ }`:

```python
name = "Kim"
score = 87.456
print(f"{name} scored {score}")       # → Kim scored 87.456
print(f"{name} scored {score:.1f}")   # → Kim scored 87.5   (:.1f = 1 decimal place)
```

> ⚠️ **Common mistake.** `"3" + 4` gives an error (`TypeError`), because Python won't add text and a number. Convert first: `int("3") + 4`.

## 2. Strings (text)

A string is text inside quotes. Each character has a position number, its **index**, **starting from 0**. You can also count from the end with negative numbers:

| Character | p | y | t | h | o | n |
|---|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 | 5 |
| From the end | -6 | -5 | -4 | -3 | -2 | -1 |

```python
s = "python"
print(s[0])     # → p    (first character)
print(s[3])     # → h
print(s[-1])    # → n    (last character)
```

A **slice** `s[start:stop]` takes the characters **from `start` up to, but not including, `stop`**. A handy check: you get `stop - start` characters.

```python
s = "python"
print(s[0:2])   # → py     positions 0 and 1 (2 characters)
print(s[2:5])   # → tho    positions 2, 3, 4 (3 characters)
print(s[2:])    # → thon   no stop = go to the end
print(s[:2])    # → py     no start = begin at 0
```

A third number is the **step**. `[::-1]` means "the whole string, walking backwards", so it reverses it:

```python
print("python"[::-1])   # → nohtyp
```

**Methods** are actions you do *to* a string. They give you back a **new** value:

| Method | What it does | Example → result |
|---|---|---|
| `.upper()` / `.lower()` | change case | `"Hi".upper()` → `"HI"` |
| `.strip()` | remove spaces at both ends | `"  hi ".strip()` → `"hi"` |
| `.replace(a, b)` | swap every `a` for `b` | `"cat".replace("c", "h")` → `"hat"` |
| `.split()` | cut into a **list** of words | `"deep learning".split()` → `["deep", "learning"]` |
| `"-".join(list)` | glue a list into one string | `"-".join(["a", "b"])` → `"a-b"` |

Two more things that work on any string: `len()` counts characters, and `in` checks whether one piece of text is inside another.

```python
s = "machine learning"
print(len(s))           # → 16   (the space counts too)
print("learn" in s)     # → True
print("LEARN" in s)     # → False  (capital letters are different characters)
```

Strings **can't be changed** once made (they are *immutable*). To "change" one, build a new string:

```python
s = "python"
try:
    s[0] = "P"                    # not allowed
except TypeError as e:
    print("Error:", e)            # → Error: 'str' object does not support item assignment
s = "P" + s[1:]                   # new string: "P" + "ython"
print(s)                          # → Python
```

> ⚠️ **Common mistake.** `"python"[6]` gives an `IndexError`. There are 6 characters, so the indices only go from 0 to 5. The last one is `s[5]`, or `s[-1]`.

## 3. Lists

A **list** holds several values in order, inside `[ ]`, separated by commas. Indexing and slicing work exactly like strings. The difference is that you **can change** a list.

```python
nums = [10, 20, 30]
print(nums[0])      # → 10
print(nums[-1])     # → 30
print(len(nums))    # → 3
```

The ways to change a list. Run them one by one and look at the list after each:

```python
nums = [10, 20, 30]
nums.append(40)       # add to the end            → [10, 20, 30, 40]
nums.insert(0, 5)     # put 5 at index 0          → [5, 10, 20, 30, 40]
nums[1] = 99          # replace the item at 1     → [5, 99, 20, 30, 40]
last = nums.pop()     # remove the last item and give it back: last = 40
print(nums, last)     # → [5, 99, 20, 30] 40
```

**Two ways to sort**, which are easy to mix up:

| Write | What happens to `a` | What you get back |
|---|---|---|
| `b = sorted(a)` | nothing, `a` stays the same | a **new** sorted list |
| `a.sort()` | `a` itself gets rearranged | `None` (nothing) |

```python
a = [3, 1, 2]
b = sorted(a)
print(a, b)        # → [3, 1, 2] [1, 2, 3]

a.sort()
print(a)           # → [1, 2, 3]
```

> ⚠️ **Common mistake.** `a = a.sort()` destroys your list. `.sort()` returns `None`, so `a` becomes `None`. Write either `a.sort()` or `a = sorted(a)`, never `a = a.sort()`.

**The copy trap.** `b = a` does **not** make a second list. It gives the *same* list a second name, so a change through `b` shows up in `a`:

```python
a = [1, 2, 3]
b = a            # b is just another name for the SAME list
b.append(4)
print(a)         # → [1, 2, 3, 4]   a changed too!
print(a is b)    # → True           (`is` asks: literally the same object?)
```

To get a real, separate copy, use `.copy()`:

```python
a = [1, 2, 3]
c = a.copy()     # a new list with the same items
c.append(4)
print(a, c)      # → [1, 2, 3] [1, 2, 3, 4]   a is safe
```

This matters later: PyTorch tensors behave the same way (`t2 = t1` doesn't copy, `t1.clone()` does).

## 4. if and for loops

An **`if`** runs some lines **only when a condition is True**. The lines that belong to it are **indented** (4 spaces). That indentation is how Python knows where the block ends.

```python
score = 72
if score >= 90:
    grade = "A"          # runs only if score >= 90
elif score >= 70:        # elif = "else if": checked only if the line above was False
    grade = "B"
else:                    # runs if nothing above was True
    grade = "C"
print(grade)             # → B
```

| Write | Means |
|---|---|
| `==` | is equal to (**two** `=` signs; one `=` stores a value) |
| `!=` | is not equal to |
| `<`  `>`  `<=`  `>=` | less than, greater than, less or equal, greater or equal |
| `and` / `or` / `not` | combine conditions: `x > 0 and x < 10` |

A **`for` loop** runs the indented lines **once for each item**. Each time round, the name after `for` holds the current item:

```python
for fruit in ["apple", "kiwi", "plum"]:
    print(fruit)
# round 1: fruit = "apple" → prints apple
# round 2: fruit = "kiwi"  → prints kiwi
# round 3: fruit = "plum"  → prints plum
```

`range()` gives a sequence of numbers to loop over. Like slicing, it **stops before** the last number:

| Write | Numbers you get |
|---|---|
| `range(4)` | 0, 1, 2, 3 |
| `range(2, 5)` | 2, 3, 4 |
| `range(0, 10, 3)` | 0, 3, 6, 9 (count by 3) |

The **running total** pattern: start at 0, then add one item per round.

```python
nums = [4, 7, 1]
total = 0
for x in nums:
    total = total + x     # round 1: 0+4=4, round 2: 4+7=11, round 3: 11+1=12
print(total)              # → 12
```

A **`while` loop** repeats *as long as* its condition is True. `break` leaves a loop immediately, and `continue` skips to the next round:

```python
n = 0
while n < 3:
    n = n + 1
    print(n)              # → 1, then 2, then 3

for i in range(5):
    if i == 2:
        continue          # skip 2
    if i == 4:
        break             # stop the loop completely
    print(i)              # → 0, 1, 3
```

In an `if`, empty things count as False: `0`, `""`, `[]`, `{}`, and `None`. Anything non-empty counts as True. So `if my_list:` means "if the list isn't empty".

> ⚠️ **Common mistake.** Writing `if x = 5:` (one `=`) is an error. To compare, use `==`. Also don't forget the **colon** `:` at the end of `if`, `for`, `while`, and `def` lines.

## 5. Tuples, unpacking, enumerate, zip

A **tuple** is like a list, but written with `( )`, and it **can't be changed**. It's used for small fixed groups, like a point `(x, y)`.

**Unpacking** puts each item of a tuple (or list) into its own variable. The number of names must match the number of items:

```python
point = (3, 4)
x, y = point        # x gets 3, y gets 4
print(x, y)         # → 3 4

a, b = 1, 2
a, b = b, a         # swap: a gets 2, b gets 1
print(a, b)         # → 2 1
```

This is how a function can give back two values: it returns one tuple, and you unpack it:

```python
def min_max(nums):
    return min(nums), max(nums)    # returns the tuple (1, 9)

low, high = min_max([4, 1, 9])
print(low, high)                   # → 1 9
```

**`enumerate`** gives you the position *and* the item while looping. **`zip`** walks through two lists side by side:

```python
for i, word in enumerate(["cat", "dog"]):
    print(i, word)          # → 0 cat, then 1 dog

for a, b in zip([1, 2, 3], [10, 20, 30]):
    print(a, b, a * b)      # → 1 10 10, then 2 20 40, then 3 30 90
```

## 6. Dictionaries and sets

A **dictionary** (`dict`) stores **pairs**: a **key** and its **value**. You look a value up **by its key**, not by position. It's written `{key: value, ...}`.

```python
ages = {"kim": 23, "lee": 25}
print(ages["kim"])     # → 23            look up by key
ages["park"] = 22      # new key: added
ages["kim"] = 24       # existing key: value replaced
print(ages)            # → {'kim': 24, 'lee': 25, 'park': 22}
```

Asking for a key that isn't there is an error. **`.get(key, default)`** returns the default instead:

```python
print(ages.get("choi", 0))   # → 0   ("choi" isn't a key, so you get the default)
print("lee" in ages)         # → True   `in` checks the KEYS
print(25 in ages)            # → False  25 is a value, not a key
```

To loop over a dict, use `.items()`. It gives you each key and value together:

```python
for name, age in ages.items():
    print(name, age)        # → kim 24, then lee 25, then park 22
```

**The counting pattern.** You'll use this constantly. Here it is traced word by word for `"a b a"`:

```python
counts = {}
for word in "a b a".split():
    counts[word] = counts.get(word, 0) + 1
print(counts)               # → {'a': 2, 'b': 1}
```

| Round | word | `counts.get(word, 0)` | counts after |
|---|---|---|---|
| 1 | `"a"` | 0 (not there yet) | `{'a': 1}` |
| 2 | `"b"` | 0 | `{'a': 1, 'b': 1}` |
| 3 | `"a"` | 1 | `{'a': 2, 'b': 1}` |

A **set** is a collection with **no duplicates** and no order. It's useful for removing repeats:

```python
print(set([1, 1, 2, 3, 3]))   # → {1, 2, 3}
s1 = {1, 2, 3}
s2 = {2, 3, 4}
print(s1 & s2)                # → {2, 3}         in both
print(s1 | s2)                # → {1, 2, 3, 4}   in either
print(s1 - s2)                # → {1}            in s1 but not s2
```

## 7. Functions

A **function** packages some code under a name so you can reuse it. The parts:

```python
def area(width, height):      # def, the name, and the inputs (parameters)
    result = width * height   # the body: indented lines that do the work
    return result             # return: the output it gives back

print(area(3, 4))             # → 12   (width=3, height=4)
big = area(10, 20)            # you can store the returned value
print(big)                    # → 200
```

**`return` vs `print`** is a very common point of confusion. `print` only *shows* something on the screen. `return` *gives the value back* so the rest of your code can use it. A function with no `return` gives back `None`:

```python
def double_print(x):
    print(x * 2)        # shows 6 on screen, but gives back nothing

def double_return(x):
    return x * 2        # gives 6 back

a = double_print(3)     # prints 6
b = double_return(3)
print(a, b)             # → None 6
```

Inputs can have **default values**, which are used if you don't pass that argument. You can also pass arguments **by name**:

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Kim"))                    # → Hello, Kim!
print(greet("Kim", "Hi"))              # → Hi, Kim!
print(greet(greeting="Hey", name="Lee"))  # → Hey, Lee!   (by name, any order)
```

**A trap to know about (a classic exam question).** A default like `bag=[]` is created **once**, when Python first reads the `def`, not every time you call the function. So every call shares the same list:

```python
def add(item, bag=[]):
    bag.append(item)
    return bag

print(add(1))     # → [1]
print(add(2))     # → [1, 2]   the same list from the first call!

def add_fixed(item, bag=None):   # the safe pattern
    if bag is None:
        bag = []                 # a fresh list on every call
    bag.append(item)
    return bag

print(add_fixed(1), add_fixed(2))   # → [1] [2]
```

A **lambda** is a one-line function with no name. You'll mostly see it as the `key=` in `sorted`, which tells `sorted` *what to sort by*:

```python
print(sorted(["bb", "a", "ccc"], key=len))     # → ['a', 'bb', 'ccc']   sort by length

students = [("kim", 88), ("lee", 95), ("park", 72)]
print(sorted(students, key=lambda s: s[1], reverse=True))
# sort by item [1] (the score), highest first → [('lee', 95), ('kim', 88), ('park', 72)]
```

> ⚠️ **Common mistake.** Forgetting `return`. If your function only `print`s, the caller gets `None`. The exercise checker will tell you "your function returned None".

## 8. Comprehensions

A **comprehension** is a one-line shortcut for "make a new list using a `for` loop". These two do exactly the same thing:

```python
# long version
squares = []
for x in range(5):
    squares.append(x * x)
print(squares)                    # → [0, 1, 4, 9, 16]

# comprehension version
squares = [x * x for x in range(5)]
print(squares)                    # → [0, 1, 4, 9, 16]
```

Read it in three parts: **[ what to put in   for name in where   if condition ]**. The `if` part is optional and filters items:

```python
nums = [1, 2, 3, 4, 5, 6]
print([x * 10 for x in nums])            # → [10, 20, 30, 40, 50, 60]
print([x for x in nums if x % 2 == 0])   # → [2, 4, 6]   keep only even numbers
```

With `{ }` you can build a dict or a set the same way:

```python
print({w: len(w) for w in ["ml", "deep"]})   # → {'ml': 2, 'deep': 4}
print({x % 3 for x in range(10)})            # → {0, 1, 2}
```

Two `for`s in one comprehension read **in the same order as two nested loops**, outer first:

```python
nested = [[1, 2], [3], [4, 5]]
flat = [x for row in nested for x in row]
# same as:  for row in nested:
#               for x in row:
#                   flat.append(x)
print(flat)                                  # → [1, 2, 3, 4, 5]
```

> ⚠️ **Common mistake.** To make a 2×3 grid of zeros, write `[[0] * 3 for _ in range(2)]`. **Don't** write `[[0] * 3] * 2`: that repeats the *same* row twice (the copy trap from section 3), so changing one row changes both.

```python
good = [[0] * 3 for _ in range(2)]
bad = [[0] * 3] * 2
good[0][0] = 1
bad[0][0] = 1
print(good)    # → [[1, 0, 0], [0, 0, 0]]
print(bad)     # → [[1, 0, 0], [1, 0, 0]]   both rows changed!
```
