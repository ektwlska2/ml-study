## 2. Strings
A string is a sequence of characters. You can **index** it (get one character) and **slice** it (get a piece).
Indexing starts at **0**. A negative index counts from the end.

```python
s = "machine learning"
s[0], s[1], s[-1]
```

**Slicing** `s[start:stop]` includes `start` but **excludes** `stop`. Leaving out a number means "from the beginning" or "to the end".

**Predict first:**

```python
s[0:7], s[8:], s[:3]
```

A third number is the **step**. A step of `-1` walks backwards, which reverses the string.

```python
s[::2], s[::-1]
```

```python
s.split()
```

## 3. Lists, and the aliasing trap
A list is an ordered collection you **can** change.

```python
a = [3, 1, 2]
a.append(5)          # add one item at the end
a
```

```python
a.extend([7, 8])     # add several items
a.insert(0, 9)       # insert 9 at position 0
a
```

```python
last = a.pop()       # remove AND return the last item
last, a
```

**`sorted(a)` vs `a.sort()`.** This difference is a common exam question.
- `sorted(a)` returns a **new** sorted list, and `a` stays the same.
- `a.sort()` sorts `a` **in place** and returns `None`.

```python
a = [3, 1, 2]
b = sorted(a)
print("a =", a, " b =", b)

result = a.sort()
print("a =", a, " result =", result)
```

### The aliasing trap
`b = a` does **not** copy the list. It puts a second name tag on the **same** list.

**Predict first:** what is `a` after the cell runs?

```python
a = [1, 2, 3]
b = a
b.append(4)
a
```

```python
a is b       # True means they are literally the same object
```

To get a real copy, use `a.copy()` or `a[:]`:

```python
a = [1, 2, 3]
c = a.copy()
c.append(4)
a, c, a is c
```

> PyTorch tensors behave the same way. `t2 = t1` doesn't copy; `t1.clone()` does. This trap will come back.

## 4. Tuples, unpacking, enumerate, zip
A **tuple** is like a list, but it can't be changed. It's often used to group a few values together.

```python
point = (3, 4)
x, y = point          # "unpacking": one name per item
x, y
```

```python
a, b = 1, 2
a, b = b, a           # swap in one line
a, b
```

A function can "return two values". Really it returns one tuple, which you unpack:

```python
def min_max(xs):
    return min(xs), max(xs)

lo, hi = min_max([4, 1, 9])
lo, hi
```

`enumerate` gives you the **index and the item** together, and `zip` walks **two lists side by side**:

```python
for i, word in enumerate(["cat", "dog", "owl"]):
    print(i, word)
```

```python
for xi, yi in zip([1, 2, 3], [10, 20, 30]):
    print(xi, yi, xi * yi)
```

## 5. Dicts and sets
A **dict** maps **keys** to **values**. You look up a value by its key, not by position.

```python
ages = {"kim": 23, "lee": 25}
ages["park"] = 22        # add a key, or overwrite it if it exists
ages
```

```python
ages["kim"]
```

Looking up a key that doesn't exist raises an error. `.get(key, default)` returns the default instead:

```python
try:
    ages["choi"]
except KeyError as e:
    print("KeyError:", e)

ages.get("choi", 0)
```

```python
"lee" in ages, 25 in ages      # `in` checks KEYS, not values
```

```python
for name, age in ages.items():
    print(name, age)
```

### The counting pattern
You'll use this one constantly, including in Diagnostic B.

```python
counts = {}
for w in "a b a c a b".split():
    counts[w] = counts.get(w, 0) + 1
counts
```

Sort the dict's items by count, from largest to smallest:

```python
sorted(counts.items(), key=lambda kv: kv[1], reverse=True)
```

A **set** is an unordered collection with **no duplicates**. It's useful for removing repeats and for set math.

```python
set([1, 1, 2, 3, 3, 3])
```

```python
s1, s2 = {1, 2, 3}, {2, 3, 4}
s1 & s2, s1 | s2, s1 - s2       # intersection, union, difference
```

## 7. Functions
`def` defines a function. Arguments can have **default values**, and you can pass arguments **by name**.

```python
def scale(xs, factor=2, shift=0):
    """Return a new list: x*factor + shift."""
    return [x * factor + shift for x in xs]

scale([1, 2]), scale([1, 2], shift=1), scale([1, 2], 10)
```

### Trap: a mutable default argument
The default list `bag=[]` is created **once**, when the function is defined, and then shared between every call.

**Predict first:** what do the two calls print?

```python
def add(item, bag=[]):
    bag.append(item)
    return bag

print(add(1))
print(add(2))
```

The correct pattern is to default to `None` and create the list inside the function:

```python
def add(item, bag=None):
    if bag is None:
        bag = []
    bag.append(item)
    return bag

print(add(1))
print(add(2))
```

A **lambda** is a tiny one-line function with no name. It's mostly used as the `key=` for sorting:

```python
sorted(["bb", "a", "ccc"], key=len)
```

```python
students = [("kim", 88), ("lee", 95), ("park", 72)]
sorted(students, key=lambda s: s[1], reverse=True)
```

## 8. Comprehensions
A **comprehension** builds a list (or dict, or set) in one line. Read it from left to right like a for-loop.

```python
squares = []
for x in range(5):
    squares.append(x * x)
squares
```

```python
[x * x for x in range(5)]       # the same thing, in one line
```

```python
[x for x in range(10) if x % 2 == 0]
```

```python
{w: len(w) for w in ["ml", "deep", "ai"]}      # dict comprehension
```

```python
{x % 3 for x in range(10)}                     # set comprehension
```

A nested comprehension reads like nested loops, **outer loop first**:

```python
nested = [[1, 2], [3], [4, 5]]
[x for row in nested for x in row]
```
