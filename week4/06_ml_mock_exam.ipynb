## I. Short answers (3 points each)

**Drill (on paper).** **Q1.** Using kNN's $k$ and a decision tree's depth as examples, explain the bias–variance trade-off. Which direction overfits?

<details><summary>Worked answer</summary>

Small $k$ and deep trees follow the training data closely: low bias, high variance (they overfit; $k = 1$ and an unlimited tree reach 100% training accuracy). Large $k$ and shallow trees smooth everything: high bias, low variance (they underfit). The best setting, chosen by cross-validation, balances the two. *(1 point for each direction correctly explained, 1 for "choose by CV".)*

</details>

**Drill (on paper).** **Q2.** Why is ridge regression the same as MAP estimation with a Gaussian prior on $w$? What does a large $\lambda$ mean as a prior belief?

<details><summary>Worked answer</summary>

MAP maximizes $\log P(\text{data} \mid w) + \log P(w)$. With Gaussian noise, the first term is $-\frac{1}{2\sigma^2}\sum_i (y_i - w^\top x_i)^2$; with the prior $w \sim \mathcal{N}(0, \tau^2 I)$, the second is $-\frac{1}{2\tau^2}\lVert w \rVert^2$. Maximizing their sum is minimizing $\sum_i (y_i - w^\top x_i)^2 + \lambda\lVert w \rVert^2$ with $\lambda = \sigma^2 / \tau^2$. A large $\lambda$ means a small $\tau^2$: a strong belief that the weights are near 0.

</details>

**Drill (on paper).** **Q3.** Bagging vs boosting: what each trains on, what kind of base model each uses, and which error each reduces.

<details><summary>Worked answer</summary>

Bagging: many **deep** trees, each on a bootstrap sample, trained independently, averaged or voted; it reduces **variance**. Boosting: **small** trees (stumps) trained one after another, each on reweighted data (AdaBoost) or on the residuals (gradient boosting); it reduces **bias**.

</details>

**Drill (on paper).** **Q4.** A defect detector is evaluated on data with 1% defects. Why is PR-AUC a better summary than ROC-AUC here?

<details><summary>Worked answer</summary>

ROC's x-axis is the false-positive rate $\frac{FP}{FP + TN}$; with 99% negatives, even many false alarms make it tiny, so ROC-AUC looks excellent. Precision $\frac{TP}{TP + FP}$ divides by the number of alarms, so false alarms are visible. Random guessing gives PR-AUC ≈ 0.01, but ROC-AUC 0.5.

</details>

**Drill (on paper).** **Q5.** Compare the hinge loss and the logistic loss as functions of $yz$. Which training points influence an SVM's solution, and which influence logistic regression's?

<details><summary>Worked answer</summary>

Hinge $\max(0, 1 - yz)$ is exactly 0 for $yz \ge 1$; logistic $\log(1 + e^{-yz})$ is positive everywhere, only approaching 0. So only points with $yz \le 1$ (on or inside the margin, or misclassified: the support vectors) affect the SVM, while **every** point pulls a little on logistic regression, with a pull of $\hat p - y$.

</details>

**Drill (on paper).** **Q6.** k-means vs a Gaussian mixture: two differences, and when k-means is a special case of the GMM.

<details><summary>Worked answer</summary>

k-means gives **hard** labels and round clusters (it uses only the distance to a centre); a GMM gives **soft** responsibilities and elliptical clusters (each component has its own covariance and weight), and it provides a density, which can score anomalies. With equal, round, very small variances, the responsibilities become 0/1, and EM becomes k-means.

</details>

**Drill (on paper).** **Q7 (8 points).** Fit $\hat y = b + wx$ by least squares to $x = (0, 1, 2, 3)$, $y = (1, 3, 2, 6)$. Give the residuals and $R^2$.

<details><summary>Worked answer</summary>

$\bar x = 1.5$, $\bar y = 3$. Deviations: $x - \bar x = (-1.5, -0.5, 0.5, 1.5)$ and $y - \bar y = (-2, 0, -1, 3)$. $S_{xy} = 3 + 0 - 0.5 + 4.5 = 7$ and $S_{xx} = 2.25 + 0.25 + 0.25 + 2.25 = 5$. So $w = \frac{7}{5} = 1.4$ and $b = 3 - 1.4 \times 1.5 = 0.9$. $\hat y = (0.9, 2.3, 3.7, 5.1)$, residuals $(0.1, 0.7, -1.7, 0.9)$, which sum to 0. $\text{SSE} = 0.01 + 0.49 + 2.89 + 0.81 = 4.2$ and $\text{SST} = 4 + 0 + 1 + 9 = 14$, so $R^2 = 1 - \frac{4.2}{14} = 0.7$. *(3 for $w$ and $b$, 2 for the residuals, 3 for $R^2$.)*

</details>

## III. Handwritten code (20 points)

**Drill (on paper).** **Q15 (10 points).** Write `kmeans(X, C, iters)` in NumPy: `X` is `(n, d)`, `C` is `(k, d)` initial centroids. Alternate assign and update for up to `iters` iterations, stop early if the labels don't change, and return `(C, labels)`. No loops over points.

<details><summary>Worked answer</summary>

```python
def kmeans(X, C, iters):
    C = C.copy()
    labels = None
    for _ in range(iters):
        D2 = ((X[:, None, :] - C[None, :, :]) ** 2).sum(axis=2)     # (n, k)
        new = D2.argmin(axis=1)                                     # (n,)
        if labels is not None and np.array_equal(new, labels):
            break
        labels = new
        C = np.array([X[labels == j].mean(axis=0) for j in range(len(C))])
    return C, labels
```

**Drill (on paper).** **Q16 (10 points).** Write `pca_fit(X, k)` returning `(mean, W)` with `W` of shape `(d, k)` holding the top-$k$ principal directions as columns (use `np.linalg.eigh`), and `pca_transform(X, mean, W)` returning the scores.

<details><summary>Worked answer</summary>

```python
def pca_fit(X, k):
    mean = X.mean(axis=0)
    Xc = X - mean
    S = Xc.T @ Xc / (len(X) - 1)                  # (d, d)
    values, vectors = np.linalg.eigh(S)            # ascending order
    order = np.argsort(values)[::-1][:k]           # the k largest
    return mean, vectors[:, order]                 # (d,), (d, k)

def pca_transform(X, mean, W):
    return (X - mean) @ W                          # (n, k)
```
