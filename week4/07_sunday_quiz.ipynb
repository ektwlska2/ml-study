**B1.** A 1-D Gaussian mixture has two components, $\mathcal{N}(2, 1)$ and $\mathcal{N}(5, 1)$, with $\pi = (0.5, 0.5)$. Compute the responsibility of component 1 for the point $x = 4$.

<details><summary>Answer</summary>

$0.5\,\mathcal{N}(4 \mid 2, 1) = 0.5 \cdot 0.399\, e^{-2} = 0.0270$ and $0.5\,\mathcal{N}(4 \mid 5, 1) = 0.5 \cdot 0.399\, e^{-0.5} = 0.1210$. So $r_1 = \frac{0.0270}{0.0270 + 0.1210} = 0.182$. The point is 2 away from component 1's centre but only 1 from component 2's.

</details>

**B2.** Points $A = (0, 0)$, $B = (0, 2)$ form one cluster; $C = (4, 0)$, $D = (4, 2)$ form the other. Compute the silhouette of $A$.

<details><summary>Answer</summary>

$a(A) = d(A, B) = 2$. $b(A) = \frac{d(A, C) + d(A, D)}{2} = \frac{4 + 4.472}{2} = 4.236$. $s = \frac{4.236 - 2}{4.236} = 0.528$.

</details>

**B3.** In gradient boosting with the squared loss and learning rate $\eta = 0.1$, the current prediction for a point is 5 and its label is 8. The next stump predicts 2.5 for it. What is the new prediction, and what is the new residual?

<details><summary>Answer</summary>

The residual before was $8 - 5 = 3$. New prediction: $5 + 0.1 \times 2.5 = 5.25$. New residual: $8 - 5.25 = 2.75$.

</details>
