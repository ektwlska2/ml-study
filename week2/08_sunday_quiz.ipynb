## Part B · On paper

**B1.** Fit a least-squares line by hand to $x = (1, 2, 3)$, $y = (2, 2, 5)$. Then compute the residuals and $R^2$.

<details><summary>Answer</summary>

$\bar x = 2$, $\bar y = 3$. Deviations: $x - \bar x = (-1, 0, 1)$ and $y - \bar y = (-1, -1, 2)$. Products: $1, 0, 2$, sum $3$. Squares: $1, 0, 1$, sum $2$. So $\hat w = 1.5$ and $\hat b = 3 - 1.5 \cdot 2 = 0$: $\hat y = 1.5x$, giving $(1.5, 3, 4.5)$. Residuals $(0.5, -1, 0.5)$, which sum to 0. $\text{SSE} = 0.25 + 1 + 0.25 = 1.5$ and $\text{SST} = 1 + 1 + 4 = 6$, so $R^2 = 1 - \frac{1.5}{6} = 0.75$.

</details>

**B2.** Logistic regression with $w = (1, 1)$, $b = 0$, sample $x = (1, -1)$, label $y = 1$, $\eta = 1$. Compute $\hat p$, the loss, and the updated $w$ and $b$.

<details><summary>Answer</summary>

$z = 1 - 1 + 0 = 0$, so $\hat p = 0.5$ and the loss is $-\ln 0.5 = 0.693$. Error $\hat p - y = -0.5$. $\nabla_w = -0.5 \times (1, -1) = (-0.5, 0.5)$ and $\partial_b = -0.5$. New $w = (1, 1) - (-0.5, 0.5) = (1.5, 0.5)$, new $b = 0.5$. Check: the new $z = 1.5 - 0.5 + 0.5 = 1.5 > 0$, so $\hat p$ went up toward the true label.

</details>

**B3.** A defect detector on 500 parts: $TP = 15$, $FN = 5$, $FP = 45$, $TN = 435$. Compute accuracy, precision, recall and $F_1$. Is this a good detector?

<details><summary>Answer</summary>

Accuracy $\frac{450}{500} = 0.90$. Precision $\frac{15}{60} = 0.25$. Recall $\frac{15}{20} = 0.75$. $F_1 = \frac{2 \cdot 0.25 \cdot 0.75}{1.0} = 0.375$. It catches most defects, but 3 out of 4 alarms are false. The 0.90 accuracy hides this, because only 4% of the parts are defective.

</details>

**B4.** Training data in 1-D: $0 \to A$, $1 \to A$, $3 \to B$, $4 \to B$, $5 \to B$. Predict $x = 2.2$ with $k = 1$ and $k = 3$.

<details><summary>Answer</summary>

Distances: $2.2, 1.2, 0.8, 1.8, 2.8$. $k = 1$: the nearest point is 3 (distance 0.8), so **B**. $k = 3$: the points at 3, 1 and 4 vote B, A, B, so **B**.

</details>

**B5.** Prior $\text{Beta}(2, 2)$ on a coin; you see 1 head in 4 flips. Write the posterior, and compute the MLE and the MAP estimate.

<details><summary>Answer</summary>

Posterior $\text{Beta}(1 + 2,\ 3 + 2) = \text{Beta}(3, 5)$. MLE $= \frac{1}{4} = 0.25$. MAP $= \frac{1 + 1}{4 + 2} = \frac{2}{6} \approx 0.333$, pulled toward 0.5 by the prior.

</details>

**B6.** One-feature ridge with no intercept: $x = (1, 2)$, $y = (2, 3)$. Compute $\hat w$ for $\lambda = 0$ and for $\lambda = 5$. Which prior belief does $\lambda = 5$ represent?

<details><summary>Answer</summary>

$\sum x_i y_i = 2 + 6 = 8$ and $\sum x_i^2 = 1 + 4 = 5$. $\lambda = 0$: $\hat w = \frac{8}{5} = 1.6$. $\lambda = 5$: $\hat w = \frac{8}{10} = 0.8$. A large $\lambda = \sigma^2 / \tau^2$ means a small prior variance $\tau^2$: a strong belief that the weight is close to 0.

</details>
