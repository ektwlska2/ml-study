## Part B · On paper

**B1.** An autoencoder `Linear(64, 32) → ReLU → Linear(32, 4)`, mirrored decoder `Linear(4, 32) → ReLU → Linear(32, 64)`. (a) Total parameters. (b) Why does it need the bottleneck (or noise) to learn anything useful? (c) What does it learn if the ReLUs are removed and the loss is the squared error on centred data?

<details><summary>Answer</summary>

(a) $64 \cdot 32 + 32 = 2080$; $32 \cdot 4 + 4 = 132$; $4 \cdot 32 + 32 = 160$; $32 \cdot 64 + 64 = 2112$; total $4484$. (b) Without it the network can copy its input and the code says nothing new; a 4-number code forces a summary of what the data has in common. (c) PCA's top-4 subspace: the decoder's columns span the top-4 eigenvectors; the codes are PCA's scores up to an invertible 4×4 change of coordinates.

</details>

**B2.** A VAE's encoder gives $\mu = (1, -1)$ and $\sigma^2 = (1, 0.5)$ for an image. (a) The KL term ($\ln 0.5 = -0.693$). (b) In one sentence each: what goes wrong without the KL term, and what goes wrong with only the KL term?

<details><summary>Answer</summary>

(a) $\frac12(1 + 1 - 1 - 0) + \frac12(0.5 + 1 - 1 + 0.693) = 0.5 + 0.597 = 1.097$. (b) Without it the clouds shrink to points ($\sigma \to 0$) and spread anywhere: an ordinary autoencoder with gaps, useless for sampling. With only it, every cloud becomes $\mathcal{N}(0, I)$ whatever the image, so the code carries no information and the decoder outputs an average digit.

</details>

**B3.** A detector for faulty bearings: 2,000 normal readings and 40 faulty ones in the test set. At its threshold it flags 60, of which 30 are faulty. (a) Precision, recall, false positive rate. (b) Its ROC-AUC is 0.95 and its average precision 0.45. Which number would you show the maintenance team, and why?

<details><summary>Answer</summary>

(a) Precision $30/60 = 0.5$; recall $30/40 = 0.75$; FPR $30/2000 = 0.015$. (b) Average precision (and precision/recall at the threshold): with 50 normal readings per faulty one, ROC's false positive rate stays small even when half the alarms are false. Precision is what the team experiences: one alarm in two is real.

</details>
