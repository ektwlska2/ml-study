## B. Dictionaries, from the start

There are only four things you do with a dict: **create** it, **look up** a key, **add or change** a key, and **loop** over it.

```python
scores = {"kim": 80, "lee": 95}      # create
print(scores["lee"])                 # → 95     look up
scores["park"] = 70                  # add (a new key)
scores["kim"] = 85                   # change (an existing key)
print(scores)                        # → {'kim': 85, 'lee': 95, 'park': 70}
```

Keys are **unique**. Assigning to a key that already exists replaces its value; it never creates a second copy.

**Looping.** A plain `for` over a dict gives you the **keys**. Use `.values()` for the values, or `.items()` for both at once:

```python
for name in scores:
    print(name)                      # → kim, then lee, then park  (keys only)

print(list(scores.values()))         # → [85, 95, 70]
print(sum(scores.values()))          # → 250

for name, score in scores.items():   # each round gives one (key, value) pair
    print(name, score)               # → kim 85, then lee 95, then park 70
```

**`key in d` searches the dict for you.** In exercise 7 you wrote a whole inner loop to look for a key. That loop is what `in` already does, and it's instant even for a million keys:

```python
d2 = {"b": 10, "c": 5}

# what you wrote (works, but long):
found = False
for j in d2:
    if j == "b":
        found = True
print(found)            # → True

# the same thing in one step:
print("b" in d2)        # → True
```

So exercise 7 becomes: copy `d1`, then go through `d2` once. Here's the trace, round by round:

```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 10, "c": 5}
merged = dict(d1)                     # a COPY of d1 (so d1 stays unchanged) → {'a': 1, 'b': 2}
for key, value in d2.items():
    if key in merged:                 # key already there → add to it
        merged[key] = merged[key] + value
    else:                             # new key → insert it
        merged[key] = value
print(merged)                         # → {'a': 1, 'b': 12, 'c': 5}
```

| Round | key, value | `key in merged`? | merged after |
|---|---|---|---|
| start |  |  | `{'a': 1, 'b': 2}` |
| 1 | `'b', 10` | yes → 2 + 10 | `{'a': 1, 'b': 12}` |
| 2 | `'c', 5` | no → insert | `{'a': 1, 'b': 12, 'c': 5}` |

There's an even shorter version that uses `.get` with a default of 0, the same idea as the counting pattern:

```python
merged = dict(d1)
for key, value in d2.items():
    merged[key] = merged.get(key, 0) + value      # missing key counts as 0
print(merged)                                     # → {'a': 1, 'b': 12, 'c': 5}
```

**The grouping pattern** (a dict whose values are lists). You'll see it a lot in data work:

```python
words = ["apple", "avocado", "banana", "blueberry", "cherry"]
groups = {}
for w in words:
    first = w[0]                      # first letter
    if first not in groups:
        groups[first] = []            # first time we see this letter: start an empty list
    groups[first].append(w)           # add the word to that letter's list
print(groups)   # → {'a': ['apple', 'avocado'], 'b': ['banana', 'blueberry'], 'c': ['cherry']}
```

| Round | w | first | groups after |
|---|---|---|---|
| 1 | apple | a | `{'a': ['apple']}` |
| 2 | avocado | a | `{'a': ['apple', 'avocado']}` |
| 3 | banana | b | `{'a': [...], 'b': ['banana']}` |
| 4 | blueberry | b | `{'a': [...], 'b': ['banana', 'blueberry']}` |
| 5 | cherry | c | `{..., 'c': ['cherry']}` |

> ⚠️ **Common mistake.** `for x in d` gives the **keys**, not the values. If you want the values, loop over `d.values()`, or use `d.items()` to get both.

## C. Loops inside loops

**The one rule:** the inner loop runs **all the way through** for **each single round** of the outer loop. Then the outer loop moves one step, and the inner loop starts again from the beginning.

```python
for i in range(2):              # outer: i = 0, then i = 1
    for j in range(3):          # inner: j = 0, 1, 2, all of it for EACH i
        print(i, j)
# → 0 0
#   0 1
#   0 2      (inner loop finished for i = 0)
#   1 0      (outer moved to i = 1; inner restarts at 0)
#   1 1
#   1 2      2 rounds × 3 rounds = 6 lines
```

You said writing it out helps, and that's the right instinct. Picture it as a grid: the outer loop picks the **row**, and the inner loop walks across the **columns** of that row.

|  | j = 0 | j = 1 | j = 2 |
|---|---|---|---|
| **i = 0** | 1st | 2nd | 3rd |
| **i = 1** | 4th | 5th | 6th |

**A trick for seeing it in code:** put a `print` with an f-string inside the loop that shows every variable. Delete it once you understand what's happening.

```python
total = 0
for i in range(1, 3):
    for j in range(1, 3):
        total = total + i * j
        print(f"i={i} j={j} i*j={i*j} total={total}")
# → i=1 j=1 i*j=1 total=1
#   i=1 j=2 i*j=2 total=3
#   i=2 j=1 i*j=2 total=5
#   i=2 j=2 i*j=4 total=9
```

**Building a list of lists** (a table) with nested loops. The new row is created **inside** the outer loop, so each row starts empty. You did this correctly in your transpose:

```python
table = []
for i in range(1, 4):            # one round per ROW
    row = []                     # a fresh, empty row for this round
    for j in range(1, 4):        # fill the row, one column at a time
        row.append(i * j)
    table.append(row)            # the row is finished: add it to the table
print(table)                     # → [[1, 2, 3], [2, 4, 6], [3, 6, 9]]
```

Watch what happens if the row is created **outside** the outer loop instead. All the rounds keep adding to the same list:

```python
table = []
row = []                         # WRONG place: only one row is ever made
for i in range(1, 3):
    for j in range(1, 3):
        row.append(i * j)
    table.append(row)
print(table)                     # → [[1, 2, 2, 4], [1, 2, 2, 4]]   the same list, added twice
```

**Your transpose, traced.** For `m = [[1, 2, 3], [4, 5, 6]]`, the outer loop picks a column `i` and the inner loop walks down the rows `j`:

| Outer round (column i) | Inner rounds (row j): picks m[j][i] | new row |
|---|---|---|
| i = 0 | m[0][0] = 1, m[1][0] = 4 | `[1, 4]` |
| i = 1 | m[0][1] = 2, m[1][1] = 5 | `[2, 5]` |
| i = 2 | m[0][2] = 3, m[1][2] = 6 | `[3, 6]` |

**Pairs without repeats.** To look at every pair *once* (and never pair an item with itself), start the inner loop one step after the outer one: `range(i + 1, n)`.

```python
nums = [2, 8, 3, 7]
n = len(nums)
for i in range(n):
    for j in range(i + 1, n):    # j always starts AFTER i
        print(nums[i], nums[j])
# → 2 8, 2 3, 2 7, 8 3, 8 7, 3 7    (6 pairs, each exactly once)
```

A loop over a dict of lists is also a nested loop: the outer loop goes over the keys, the inner loop over that key's list.

```python
groups = {"a": ["apple", "avocado"], "b": ["banana"]}
for letter, words in groups.items():
    for w in words:
        print(letter, w)         # → a apple, then a avocado, then b banana
```

**2. Don't name a variable `list`** (or `sum`, `max`, `dict`, `str`). Python lets you, but then the built-in stops working until you restart:

```python
list = [1, 2, 3]                 # this hides the built-in list()
try:
    print(list("abc"))
except TypeError as e:
    print("Error:", e)           # → Error: 'list' object is not callable
del list                         # remove our variable, so the built-in comes back
print(list("abc"))               # → ['a', 'b', 'c']
```

Use names that say what the thing is: `result`, `out`, `words`, `rows`.

**3. `if condition: return True` / `else: return False` is just `return condition`.** A comparison already *is* True or False:

```python
def is_even_long(n):
    if n % 2 == 0:
        return True
    else:
        return False

def is_even(n):
    return n % 2 == 0            # the comparison is already True or False

print(is_even_long(4), is_even(4), is_even(7))   # → True True False
```

**4. In fizzbuzz, check the "both" case first.** Then the other two conditions become simpler, because they don't need `and ... != 0`:

```python
def fizzbuzz(n):
    out = []
    for i in range(1, n + 1):
        if i % 15 == 0:          # divisible by both 3 and 5: check this FIRST
            out.append("FizzBuzz")
        elif i % 3 == 0:         # only reached when it's not divisible by 15
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        else:
            out.append(str(i))
    return out

print(fizzbuzz(15)[-3:])         # → ['13', '14', 'FizzBuzz']
```
