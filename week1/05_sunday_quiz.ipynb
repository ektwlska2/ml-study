## Part B · Math on paper

**B1.** A factory has two machines. Machine 1 makes 70% of the parts and 2% of its parts are defective. Machine 2 makes 30% and 10% of its parts are defective. A part is defective. What is the probability that it came from machine 2?

<details><summary>Answer</summary>

P(def and M1) = 0.7 × 0.02 = 0.014. P(def and M2) = 0.3 × 0.10 = 0.03. P(M2 | def) = 0.03 / 0.044 ≈ **0.682**.

</details>

**B2.** You observe 4 successes in 20 trials. Write the log-likelihood, take its derivative, and find the MLE of p. Then give the MAP estimate with a Beta(2, 2) prior.

<details><summary>Answer</summary>

log L = 4 log p + 16 log(1 − p). Derivative: 4/p − 16/(1 − p) = 0 → **p = 0.2**. MAP: (4 + 1)/(20 + 2) = 5/22 ≈ **0.227**.

</details>

**B3.** Find the eigenvalues of [[5, 2], [2, 2]].

<details><summary>Answer</summary>

det = (5 − λ)(2 − λ) − 4 = λ² − 7λ + 6 = (λ − 6)(λ − 1). So **λ = 6 and 1**.

</details>

**B4.** L(w) = (w − 5)², starting at w = 1 with learning rate 0.25. Do two gradient-descent steps.

<details><summary>Answer</summary>

L' = 2(w − 5). Step 1: w = 1 − 0.25 × (−8) = **3**. Step 2: w = 3 − 0.25 × (−4) = **4**.

</details>

**B5.** A 3-class model predicts q = [0.2, 0.7, 0.1] and the true class is 1. What is the cross-entropy loss? What would it be if the true class were 2? Which is bigger, and why?
