## Part B · Math on paper

**B1.** A factory has two machines. Machine 1 ($M_1$) makes 70% of the parts and 2% of its parts are defective. Machine 2 ($M_2$) makes 30% and 10% of its parts are defective. A part is defective ($D$). What is $P(M_2 \mid D)$?

<details><summary>Answer</summary>

$P(D \cap M_1) = 0.7 \times 0.02 = 0.014$. $P(D \cap M_2) = 0.3 \times 0.10 = 0.03$. $P(M_2 \mid D) = \frac{0.03}{0.014 + 0.03} = \frac{0.03}{0.044} \approx \mathbf{0.682}$.

</details>

**B2.** You observe 4 successes in 20 trials. Write the log-likelihood $\log L(p)$, take its derivative, and find the MLE $\hat p$. Then give the MAP estimate with a $\text{Beta}(2, 2)$ prior.

<details><summary>Answer</summary>

$\log L = 4 \log p + 16 \log(1 - p)$. Derivative: $\frac{4}{p} - \frac{16}{1 - p} = 0 \Rightarrow \hat p = \mathbf{0.2}$. MAP: $\frac{4 + 1}{20 + 2} = \frac{5}{22} \approx \mathbf{0.227}$.

</details>

**B3.** Find the eigenvalues of $\begin{pmatrix} 5 & 2 \\ 2 & 2 \end{pmatrix}$.

<details><summary>Answer</summary>

$\det(A - \lambda I) = (5 - \lambda)(2 - \lambda) - 4 = \lambda^2 - 7\lambda + 6 = (\lambda - 6)(\lambda - 1)$. So $\lambda = \mathbf{6}$ and $\mathbf{1}$.

</details>

**B4.** $L(w) = (w - 5)^2$, starting at $w = 1$ with learning rate $\eta = 0.25$. Do two gradient-descent steps.

<details><summary>Answer</summary>

$L'(w) = 2(w - 5)$. Step 1: $w = 1 - 0.25 \times (-8) = \mathbf{3}$. Step 2: $w = 3 - 0.25 \times (-4) = \mathbf{4}$.

</details>

**B5.** A 3-class model predicts $q = (0.2, 0.7, 0.1)$ and the true class is 1. What is the cross-entropy loss? What would it be if the true class were 2? Which is bigger, and why?

<details><summary>Answer</summary>
