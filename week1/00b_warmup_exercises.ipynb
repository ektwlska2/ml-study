## 1. Invert a dict
Swap keys and values. You can assume all values are different.

`invert({"a": 1, "b": 2})` → `{1: "a", 2: "b"}`

```python
def invert(d):
    inverted = {}
    for key,value in d.items():
        inverted[value] = key
    return inverted

print(invert({"a": 1, "b": 2}))
```

## 2. Group words by length
Return a dict mapping each length to the list of words with that length, in their original order.

`group_by_length(["hi", "ok", "cat", "a"])` → `{2: ["hi", "ok"], 3: ["cat"], 1: ["a"]}`. Use the grouping pattern.

```python
def group_by_length(words):
    grouped = {}
    for i in words:
        if len(i) not in grouped:
            grouped[len(i)] = []
        grouped[len(i)].append(i)
    return grouped

print(group_by_length(["hi", "ok", "cat", "a"]))
```

## 3. Keys in both
Return a **sorted list** of the keys that appear in both dicts.

`common_keys({"a": 1, "b": 2, "c": 3}, {"c": 0, "a": 9, "z": 1})` → `["a", "c"]`. Use `in`, not a second loop.

```python
def common_keys(d1, d2):
    lst = []
    for k in d1.keys():
        if k in d2:
            lst.append(k)
    return lst

print(common_keys({"a": 1, "b": 2, "c": 3}, {"c": 0, "a": 9, "z": 1}))
```

## 4. Count pairs that sum to a target
Count the pairs of **positions** `i < j` where `nums[i] + nums[j] == target`.

`count_pairs([2, 8, 3, 7, 5, 5], 10)` → `3` (2+8, 3+7, 5+5). And `count_pairs([1, 1, 1], 2)` → `3`, because the positions (0,1), (0,2), (1,2) are 3 different pairs.

```python
def count_pairs(nums, target):
    l = []
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j] == target:
                l.append((i,j))
    return len(l) 

print(count_pairs([2, 8, 3, 7, 5, 5], 10),count_pairs([1, 1, 1], 2))
```

## 5. Add two matrices
Two lists of lists with the same shape. Return a new one where each entry is the sum of the matching entries.

`matrix_add([[1, 2], [3, 4]], [[10, 20], [30, 40]])` → `[[11, 22], [33, 44]]`. Use a nested loop, and make each new row inside the outer loop.

```python
def matrix_add(a, b):
    if len(a) != len(b) or len(a[0]) != len(b[0]):
        return "error the matrix size must match"
     
    new = []
    for i in range(len(a)):
        new.append([])
        for j in range(len(a[0])):
            new[i].append(a[i][j] + b[i][j])
    return new

print(matrix_add([[1, 2], [3, 4]], [[10, 20], [30, 40]]))
```

## 6. Average score per student
The input maps each name to a list of scores. Return a dict mapping each name to the average (a float), or `None` if the list is empty.

`averages({"kim": [80, 90], "lee": [70, 75, 77]})` → `{"kim": 85.0, "lee": 74.0}`

```python
def averages(scores):
    grp = {}
    for k,v in scores.items():
        if len(v) == 0 :
            grp[k] = None

        else :
            grp[k] = sum(v)/len(v)
    return grp

print(averages({"kim": [80, 90],"park":[], "lee": [70, 75, 77]}))
```
